```python
# ==== PASO P-1: Instalación de librerías y setup ====
!pip -q install --upgrade pip
!pip -q install pandas numpy scipy statsmodels linearmodels patsy \
               matplotlib scikit-learn openpyxl xlsxwriter pyreadstat tabulate

import warnings, os
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from google.colab import files
from tabulate import tabulate
from scipy import stats
import statsmodels.api as sm
from linearmodels.panel import PanelOLS, RandomEffects, PooledOLS

plt.rcParams['figure.figsize'] = (7.5, 4.5)
plt.rcParams['axes.grid'] = True
plt.rcParams['figure.dpi'] = 125


# ==== PASO P-2: Subir y revisar tu archivo Excel ====
uploaded = files.upload()  # Selecciona el Excel desde tu PC
file_name = list(uploaded.keys())[0]
print(f"\nArchivo subido correctamente: {file_name}")

# Leer primera hoja
df = pd.read_excel(file_name, sheet_name=0)

# Info básica
print(f"\nFilas: {df.shape[0]} | Columnas: {df.shape[1]}")
print("\nNombres de columnas:\n", list(df.columns))

# Vista previa
display(df.head(10))


# ==== PASO P-3: Preparar el panel (nombres, logs, rezagos, filtros) ====

# 1) Renombrar columnas a estándar (ajusta si fuera necesario)
rename_map = {
    'Country': 'country',
    'Year': 'year',
    'Yit_PIB_PC': 'y',
    'Xit_DENS_VIAS': 'x1',
    'X2it_INT_UTI': 'x2_int',
    'Z1it_EDU': 'z1',
    'Z2it_FBCF': 'z2',
    'Z3it_I+D': 'z3',
    'Z4it_DEN_POB': 'z4'
}
df_std = df.rename(columns=rename_map).copy()

# 2) Ordenar por país-año
df_std = df_std.sort_values(['country','year']).reset_index(drop=True)

# 3) Crear logs (NaN si ≤ 0 o no numérico)
for c in ['y','x1','x2_int','z1','z2','z3','z4']:
    num = pd.to_numeric(df_std[c], errors='coerce')
    df_std[f'ln_{c}'] = np.where(num > 0, np.log(num), np.nan)

# 4) Rezagos t-1 para x2_int y z2 por país
df_std[['l1_ln_x2_int','l1_ln_z2']] = (
    df_std.groupby('country')[['ln_x2_int','ln_z2']].shift(1)
)

# 5) Filtrar base válida para el modelo
cols_req = ['ln_y','ln_x1','l1_ln_x2_int','ln_z1','l1_ln_z2','ln_z3','ln_z4']
n0 = len(df_std)
panel = df_std.dropna(subset=cols_req).copy()
n_kept = len(panel)
print(f"Observaciones iniciales: {n0} | válidas para el modelo: {n_kept} | filtradas: {n0 - n_kept}")

# 6) Cobertura por país (nº de observaciones)
coverage = panel.groupby('country', as_index=False).size().rename(columns={'size':'n_obs'})
coverage = coverage.sort_values('n_obs', ascending=False)
print("\nCobertura por país (tras filtros):")
print(tabulate(coverage, headers='keys', tablefmt='psql', showindex=False))

# 7) Vista previa final
display(panel.head(10))


# ==== PASO P-4A — (KIT) Estimar FE two-way con clúster y recrear res_fe si falta ====
from linearmodels.panel import PanelOLS

# Comprobaciones mínimas
if 'df_std' not in globals():
    raise RuntimeError("df_std no existe. Ejecuta P-2 y P-3 antes de este bloque.")

cols_req_full = cols_req + ['country','year']
faltan = [c for c in cols_req_full if c not in df_std.columns]
if faltan:
    raise RuntimeError(f"Faltan columnas en df_std: {faltan}. Revisa el renombrado en P-3.")

if 'panel' not in globals() or panel is None or panel.empty:
    panel = df_std.dropna(subset=cols_req_full).copy()
if panel.empty:
    raise RuntimeError("panel está vacío tras filtros. Revisa ceros/NaN en logs o relaja filtros puntualmente.")

# Índice panel y modelo
panel_idx = panel.set_index(['country','year']).copy()
formula_fe = 'ln_y ~ l1_ln_x2_int + ln_x1 + ln_z1 + l1_ln_z2 + ln_z3 + ln_z4 + EntityEffects + TimeEffects'
mod_fe = PanelOLS.from_formula(formula_fe, data=panel_idx)

# Ajuste con clúster por país
res_fe = mod_fe.fit(cov_type='clustered', cluster_entity=True)

print("✅ FE two-way estimado.")
print(f"Obs = {res_fe.nobs} | Países = {res_fe.entity_info.total} | Años = {res_fe.time_info.total}\n")
print(res_fe)



# ==== PASO P-4A.1 — Pesaran CD sobre residuos FE (robusto) ====
# Requiere res_fe creado en P-4A
resid = res_fe.resids
if not isinstance(resid.index, pd.MultiIndex):
    raise RuntimeError("Los residuos no tienen MultiIndex (entity,time). Revisa que panel_idx use ['country','year'].")

resid = resid.rename("u").reset_index().rename(columns={"entity":"country","time":"year"})
resid['year'] = pd.to_numeric(resid['year'], errors='coerce')

U = resid.pivot_table(index="year", columns="country", values="u").dropna(axis=1, how='all')
n = U.shape[1]
if n < 2:
    print("Pesaran CD: n<2 países con datos; no se calcula.")
else:
    corr = U.corr(min_periods=3)
    mask = np.triu(np.ones((n,n), dtype=bool), k=1)
    rhos = corr.where(mask).stack()
    if rhos.empty:
        print("Pesaran CD: sin pares con ≥3 años comunes.")
    else:
        rho_bar = rhos.mean()
        def common_T(i,j): return U[[i,j]].dropna().shape[0]
        pairs = list(rhos.index)
        T_bar = float(np.mean([common_T(i,j) for i,j in pairs])) if pairs else np.nan
        CD = np.sqrt(2*T_bar/(n*(n-1))) * rhos.sum() if (n>1 and len(rhos)>0 and (T_bar or 0)>0) else np.nan
        print(f"n países = {n}, T_efectivo ≈ {T_bar:.1f}")
        print(f"rhō = {rho_bar:.4f}")
        print(f"Pesaran CD ≈ {CD:.3f}  ->  |CD|>1.96 sugiere dependencia (5%)")


# ==== PASO P-4B: Pooled OLS y Random Effects (RE) ====
panel_idx = panel.set_index(['country','year']).copy()

# Pooled OLS con constante
X_pool = panel_idx[['l1_ln_x2_int','ln_x1','ln_z1','l1_ln_z2','ln_z3','ln_z4']]
X_pool_c = sm.add_constant(X_pool)
y_pool = panel_idx['ln_y']
res_pool = PooledOLS(y_pool, X_pool_c).fit(cov_type='robust')
print("\n=== Pooled OLS (robust) ===\n", res_pool)

# Random Effects
res_re = RandomEffects.from_formula(
    'ln_y ~ l1_ln_x2_int + ln_x1 + ln_z1 + l1_ln_z2 + ln_z3 + ln_z4',
    data=panel_idx
).fit()
print("\n=== Random Effects (RE) ===\n", res_re)

# Comparación rápida
compare = pd.concat([
    res_pool.params.rename('Pooled'),
    res_re.params.rename('RE'),
    res_fe.params.rename('FE_TWFE')
], axis=1)
print("\n=== Comparación rápida de coeficientes ===\n")
print(compare)


# ==== PASO P-4C: Hausman FE vs RE ====
b_fe = res_fe.params
b_re = res_re.params
common = b_fe.index.intersection(b_re.index)
b_diff = (b_fe[common] - b_re[common]).values

V_fe = res_fe.cov.loc[common, common].values
V_re = res_re.cov.loc[common, common].values
V_diff = V_fe - V_re
V_inv  = np.linalg.pinv(V_diff)

hausman_stat = float(b_diff.T @ V_inv @ b_diff)
df_h = len(common)
p_h  = 1 - stats.chi2.cdf(hausman_stat, df_h)

print("=== Hausman test (FE vs RE) ===")
print(f"Chi2({df_h}) = {hausman_stat:0.3f}  |  p-value = {p_h:0.4f}")
print("Regresores comunes:", list(common))


# ==== PASO P-5: Coefplot (FE clúster) + exportar tabla FE ====
order = ["l1_ln_x2_int","ln_x1","ln_z1","l1_ln_z2","ln_z3","ln_z4"]

fe_table = pd.DataFrame({
    "coef": res_fe.params,
    "std_err": res_fe.std_errors,
    "t": res_fe.tstats,
    "p_value": res_fe.pvalues,
}).reindex(order)

# Coefplot con IC95%
ci95 = 1.96 * fe_table["std_err"].values
ypos = np.arange(len(fe_table))

plt.figure()
plt.errorbar(fe_table["coef"].values, ypos, xerr=ci95, fmt='o', capsize=3)
plt.yticks(ypos, fe_table.index)
plt.axvline(0, linestyle='--')
plt.xlabel("Estimación (FE, clúster país) | IC 95%")
plt.title("Coeficientes — Modelo FE (Two-Way)")
plt.tight_layout()
plt.savefig("coefplot_FE.png", dpi=300)
plt.show()

fe_table.round(4).to_csv("tabla_FE_TWFE.csv")
print("\nGuardado: coefplot_FE.png, tabla_FE_TWFE.csv")


# ==== PASO P-5.1: R² within/between/overall y dimensiones ====
r2_tbl = pd.DataFrame({
    "Métrica": ["R2_within","R2_between","R2_overall","n_entity","n_time","n_obs"],
    "Valor": [
        getattr(res_fe, "rsquared_within", np.nan),
        getattr(res_fe, "rsquared_between", np.nan),
        getattr(res_fe, "rsquared_overall", np.nan),
        res_fe.entity_info.total,
        res_fe.time_info.total,
        res_fe.nobs
    ]
}).round(4)
print("\n=== R² y dimensiones del modelo FE ===\n")
print(tabulate(r2_tbl, headers='keys', tablefmt='psql', showindex=False))
r2_tbl.to_csv("tabla_R2_y_dimensiones_FE.csv", index=False)
print("Guardado: tabla_R2_y_dimensiones_FE.csv")

# ==== PASO P-5.2: VIF within (two-way demeaning) ====
from statsmodels.api import OLS, add_constant

def twoway_demean(df, entity_col, time_col, cols):
    out = df.copy()
    g = out.groupby(entity_col)[cols].transform('mean')
    t = out.groupby(time_col)[cols].transform('mean')
    overall = out[cols].mean()
    out_dm = out[cols] - g - t + overall
    out_dm[entity_col] = out[entity_col]
    out_dm[time_col]  = out[time_col]
    return out_dm

reg_cols = order.copy()
X = panel[["country","year"] + reg_cols].dropna().copy()
X_dm = twoway_demean(X, "country", "year", reg_cols).dropna()

vif_rows = []
for c in reg_cols:
    y = X_dm[c].values
    Xo = add_constant(X_dm[[col for col in reg_cols if col != c]].values)
    r2 = OLS(y, Xo).fit().rsquared
    vif = 1.0/(1.0 - r2) if r2 < 0.9999 else np.inf
    vif_rows.append({"variable": c, "R2_aux": round(r2,3), "VIF_within": round(vif,3)})

vif_within = pd.DataFrame(vif_rows).sort_values("VIF_within", ascending=False)
print("\n=== VIF within (two-way demeaning) ===\n")
print(tabulate(vif_within, headers='keys', tablefmt='psql', showindex=False))
vif_within.to_csv("A3_VIF_within.csv", index=False)
print("Guardado: A3_VIF_within.csv")


# ==== PASO P-6: FE con Driscoll–Kraay (robusto a dependencia transversal) ====
res_fe_dk = mod_fe.fit(cov_type='kernel', kernel='bartlett', bandwidth=3)
print(res_fe_dk)

comp_dk = pd.DataFrame({
    "FE_cluster_coef": res_fe.params.reindex(order),
    "FE_cluster_se": res_fe.std_errors.reindex(order),
    "FE_DK_coef": res_fe_dk.params.reindex(order),
    "FE_DK_se": res_fe_dk.std_errors.reindex(order)
}).round(4)
print("\n=== FE (clúster) vs FE (Driscoll–Kraay) ===\n")
print(tabulate(comp_dk, headers='keys', tablefmt='psql', showindex=True))
comp_dk.to_csv("A4_FE_cluster_vs_DK.csv")
print("Guardado: A4_FE_cluster_vs_DK.csv")


# ==== PASO P-6C — FE TWFE con Driscoll–Kraay (cross-sectional dependence) ====

from linearmodels.panel import PanelOLS

# Reutilizamos el mismo diseño del P-4A
panel_idx = panel.set_index(['country','year']).copy()

mod_fe_DK = PanelOLS.from_formula(
    'ln_y ~ l1_ln_x2_int + ln_x1 + ln_z1 + l1_ln_z2 + ln_z3 + ln_z4 + EntityEffects + TimeEffects',
    data=panel_idx
)

# Driscoll–Kraay: cov_type='kernel' (Bartlett), bandwidth=3
res_fe_DK = mod_fe_DK.fit(cov_type='kernel', kernel='bartlett', bandwidth=3)
print("\n=== FE dos-vías con Driscoll–Kraay (Bartlett, bw=3) ===\n")
print(res_fe_DK)

# Tabla compacta para comparar con FE_cluster
dk_table = (pd.DataFrame({
    'coef_DK': res_fe_DK.params,
    'se_DK':   res_fe_DK.std_errors,
    'p_DK':    res_fe_DK.pvalues
})
 .loc[['l1_ln_x2_int','ln_x1','ln_z1','l1_ln_z2','ln_z3','ln_z4']]
 .round(4)
)
print("\nTabla DK (coef, se, p):\n", dk_table)

# Exporta por si quieres pegarla en Word
dk_table.to_csv("tabla_FE_DK.csv")
print("\nArchivo exportado: tabla_FE_DK.csv")

#Micro-robustez 1 — Sensibilidad de bandwidth en Driscoll–Kraay
# ==== PASO P-6C.1 — Sensibilidad Driscoll–Kraay (bw = 2, 3, 4) ====

from linearmodels.panel import PanelOLS
import pandas as pd

panel_idx = panel.set_index(['country','year']).copy()

mod_fe_DK_bw = PanelOLS.from_formula(
    'ln_y ~ l1_ln_x2_int + ln_x1 + ln_z1 + l1_ln_z2 + ln_z3 + ln_z4 + EntityEffects + TimeEffects',
    data=panel_idx
)

rows = []
for bw in [2, 3, 4]:
    res = mod_fe_DK_bw.fit(cov_type='kernel', kernel='bartlett', bandwidth=bw)
    for v in ['l1_ln_x2_int','ln_x1','ln_z1','l1_ln_z2','ln_z3','ln_z4']:
        rows.append({
            'variable': v,
            'bandwidth': bw,
            'coef': res.params.get(v, float('nan')),
            'se': res.std_errors.get(v, float('nan')),
            'p': res.pvalues.get(v, float('nan'))
        })

dk_bw = pd.DataFrame(rows)
dk_bw = dk_bw[['variable','bandwidth','coef','se','p']].round(4)
print("\n=== Sensibilidad DK por bandwidth (bw=2,3,4) ===\n", dk_bw)

# Tabla “bonita” solo para la variable clave + exportación para Apéndice
dk_bw_key = dk_bw[dk_bw['variable']=='l1_ln_x2_int'].copy()
dk_bw_key.to_csv("A1_FE_DK_bandwidth_sensitivity.csv", index=False)
print("\nArchivo exportado: A1_FE_DK_bandwidth_sensitivity.csv")


# ==== P-6D.0 — Chequeo de 'panel' y tipado básico ====
import pandas as pd
import numpy as np

req = ['country','year','ln_y','l1_ln_x2_int','ln_x1','ln_z1','l1_ln_z2','ln_z3','ln_z4']
assert 'panel' in globals(), "No existe 'panel' en memoria. Ejecuta P-3 (preparación)."

missing = [c for c in req if c not in panel.columns]
assert not missing, f"Faltan columnas en 'panel': {missing}. Repite P-3."

# Copia y sanea tipos
base = panel[req].copy()
base['country'] = base['country'].astype(str)
base['year'] = pd.to_numeric(base['year'], errors='coerce').astype('Int64')
for c in req:
    if c not in ['country','year']:
        base[c] = pd.to_numeric(base[c], errors='coerce')

# Elimina filas sin año/país
base = base.dropna(subset=['country','year']).copy()
base['year'] = base['year'].astype(int)

print("Panel OK. Filas:", len(base), "| Años:", sorted(base['year'].unique()))

# ==== P-6D.A — CCE-FE mínimo: c-averages de ln_y y l1_ln_x2_int ====
from linearmodels.panel import PanelOLS

df = base.copy().sort_values(['year','country'])

# Promedios transversales (solo Y y X principal)
cavg_mini = df.groupby('year')[['ln_y','l1_ln_x2_int']].transform('mean')
cavg_mini.columns = ['cavg_ln_y','cavg_l1_ln_x2_int']
df = pd.concat([df, cavg_mini], axis=1)

# Construye regresores (baseline + c-averages mini) — sin L1
rhs = ['l1_ln_x2_int','ln_x1','ln_z1','l1_ln_z2','ln_z3','ln_z4','cavg_ln_y','cavg_l1_ln_x2_int']
need = ['country','year','ln_y'] + rhs
df = df.dropna(subset=need).copy()

# Índice e insumos
df = df.set_index(['country','year']).sort_index()
Y = df['ln_y']
X = df[rhs].replace([np.inf,-np.inf], np.nan).dropna(how='any')
Y = Y.loc[X.index]

# Quita países con <2 observaciones
keep = X.reset_index().groupby('country')['year'].size()
keep = keep[keep>=2].index
X = X.loc[X.index.get_level_values('country').isin(keep)]
Y = Y.loc[X.index]

print("CCE mínimo — obs:", len(Y), "| países:", len(keep))

# Estima FE (entity) sin efectos de tiempo; inferencia por clúster país
res_cce_min = PanelOLS(Y, X, entity_effects=True, time_effects=False)\
                .fit(cov_type='clustered', cluster_entity=True)
print("\n=== FE (CCE mínimo: cavg de Y y X principal) ===")
print(res_cce_min)

# Tabla ejecutiva
sel = ['l1_ln_x2_int','ln_x1','ln_z1','l1_ln_z2','ln_z3','ln_z4']
tbl_cce_min = pd.DataFrame({
    'coef_CCE': res_cce_min.params.reindex(sel),
    'se_CCE':   res_cce_min.std_errors.reindex(sel),
    'p_CCE':    res_cce_min.pvalues.reindex(sel)
}).round(4)
print("\nTabla CCE mínimo (coef, se, p):\n", tbl_cce_min)

tbl_cce_min.to_csv("A2_FE_CCE_min.csv")
print("Exportado: A2_FE_CCE_min.csv")


# ==== PASO P-7: Robustez A — Excluir países con n_obs < 8 ====
valid_c = coverage.loc[coverage['n_obs']>=8, 'country']
panel_rA = panel[panel['country'].isin(valid_c)].copy()

res_fe_rA = PanelOLS.from_formula(
    'ln_y ~ l1_ln_x2_int + ln_x1 + ln_z1 + l1_ln_z2 + ln_z3 + ln_z4 + EntityEffects + TimeEffects',
    data=panel_rA.set_index(['country','year'])
).fit(cov_type='clustered', cluster_entity=True)

print(res_fe_rA)


# ==== PASO P-8: Robustez B — Retardo t-2 de intermodalidad ====
df_std['l2_ln_x2_int'] = df_std.groupby('country')['ln_x2_int'].shift(2)
panel_l2 = df_std.dropna(subset=['ln_y','ln_x1','l2_ln_x2_int','ln_z1','l1_ln_z2','ln_z3','ln_z4']).copy()

res_fe_l2 = PanelOLS.from_formula(
    'ln_y ~ l2_ln_x2_int + ln_x1 + ln_z1 + l1_ln_z2 + ln_z3 + ln_z4 + EntityEffects + TimeEffects',
    data=panel_l2.set_index(['country','year'])
).fit(cov_type='clustered', cluster_entity=True)

print(res_fe_l2)


# ==== PASO P-9: Summaries completos + Tabla ejecutiva (FE/RE/Pooled/DK) ====
def save_summary(model_result, basename):
    with open(f"{basename}.txt","w",encoding="utf-8") as f:
        f.write(model_result.summary.as_text())
    try:
        html = model_result.summary.as_html()
        with open(f"{basename}.html","w",encoding="utf-8") as f:
            f.write(html)
    except Exception:
        pass

save_summary(res_fe,    "summary_FE_TWFE")
save_summary(res_fe_dk, "summary_FE_TWFE_DK")
save_summary(res_re,    "summary_RE")
save_summary(res_pool,  "summary_Pooled")

def starify(p):
    return "***" if p<0.01 else ("**" if p<0.05 else ("*" if p<0.10 else ""))

def tidy(res, label, order):
    df = pd.DataFrame({"coef": res.params, "se": res.std_errors, "p": res.pvalues}).reindex(order)
    df[label] = df["coef"].round(3).astype(str) + " (" + df["se"].round(3).astype(str) + ")" + df["p"].apply(starify)
    return df[[label]]

exec_tbl = tidy(res_fe, "FE (clúster país)", order).join(
           tidy(res_fe_dk, "FE (Driscoll–Kraay)", order)).join(
           tidy(res_re, "RE", order), how="left").join(
           tidy(res_pool, "Pooled (robust)", order), how="left")

print("\n=== Tabla ejecutiva de coeficientes ===\n")
print(tabulate(exec_tbl, headers='keys', tablefmt='psql', showindex=True))
exec_tbl.to_csv("tabla_ejecutiva_coeficientes.csv")
print("Guardados: summaries *.txt/*.html y tabla_ejecutiva_coeficientes.csv")

# --- PASO P-10: TEST DE CSD POST-CCE ---
# (Requiere 'res_cce_min' del Paso P-6D.A)

print("\n=== Test de Pesaran CD sobre Residuos CCE (Post-Estimación) ===")

# 1. Obtener residuos del CCE
resid_cce = res_cce_min.resids

# 2. Re-formatear residuos
resid_cce_df = resid_cce.rename("u").reset_index().rename(columns={"entity":"country", "time":"year"})
resid_cce_df['year'] = pd.to_numeric(resid_cce_df['year'], errors='coerce')
U_cce = resid_cce_df.pivot_table(index="year", columns="country", values="u").dropna(axis=1, how='all')

n_cce = U_cce.shape[1]

if n_cce < 2:
    print("Pesaran CD (Post-CCE): n<2 países, no se calcula.")
else:
    # 3. Calcular Correlaciones de Residuos CCE
    corr_cce = U_cce.corr(min_periods=3)
    mask_cce = np.triu(np.ones((n_cce, n_cce), dtype=bool), k=1)
    rhos_cce = corr_cce.where(mask_cce).stack()

    if rhos_cce.empty:
        print("Pesaran CD (Post-CCE): Sin pares con >=3 años comunes.")
    else:
        # 4. Calcular Test CD
        def common_T(i, j, U_df): return U_df[[i,j]].dropna().shape[0]
        pairs_cce = list(rhos_cce.index)
        T_bar_cce = float(np.mean([common_T(i, j, U_cce) for i, j in pairs_cce]))

        CD_stat_cce = np.sqrt(2 * T_bar_cce / (n_cce * (n_cce - 1))) * rhos_cce.sum()
        p_value_cce = 2 * (1 - stats.norm.cdf(abs(CD_stat_cce))) # P-valor de dos colas

        print(f"n países (CCE) = {n_cce}, T_efectivo (CCE) ≈ {T_bar_cce:.1f}")
        print(f"Estadístico CD (Post-CCE) = {CD_stat_cce:.3f}")
        print(f"P-valor (Post-CCE) = {p_value_cce:.4f}")


#GRÁFICOS DE IMPORTANCIA

# 1. Evolución temporal con leyenda de países
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))
for c in sorted(panel['country'].unique()):
    tmp = panel.loc[panel['country']==c].sort_values('year')
    plt.plot(tmp['year'], tmp['ln_y'], alpha=0.7, linewidth=1.5, label=str(c))

plt.title("Evolución del PIB per cápita (ln_y) por país — UE-27")
plt.xlabel("Año")
plt.ylabel("ln(PIB per cápita, PPA)")
plt.tight_layout()

# Leyenda fuera del gráfico para que no tape líneas
plt.legend(title="País", bbox_to_anchor=(1.02, 1), loc='upper left', frameon=False)
plt.tight_layout()
plt.savefig("grafico_evolucion_ln_y_leyenda.png", dpi=300, bbox_inches='tight')
plt.show()


#3. Comparación visual de coeficientes entre modelos
modelos = {
    'Pooled': res_pool.params,
    'RE': res_re.params,
    'FE': res_fe.params
}

coef_df = pd.DataFrame(modelos).reindex(order)
coef_df.plot(kind='barh', figsize=(8,5))
plt.axvline(0, color='black', linestyle='--')
plt.title("Comparación de coeficientes entre modelos")
plt.xlabel("Elasticidad estimada")
plt.tight_layout()
plt.savefig("grafico_comparacion_modelos.png", dpi=300)
plt.show()

#4. Test de dependencia transversal (mapa de correlaciones de residuos)
import seaborn as sns
corr = U.corr()
plt.figure(figsize=(9,7))
sns.heatmap(corr, cmap="coolwarm", center=0)
plt.title("Correlaciones cruzadas de residuos (dependencia transversal)")
plt.tight_layout()
plt.savefig("grafico_dependencia_transversal.png", dpi=300)
plt.show()

#5. Análisis de robustez gráfica
robust = pd.DataFrame({
    'Modelo': ['Base FE', 'Robustez A', 'Robustez B'],
    'Coeficiente': [res_fe.params['l1_ln_x2_int'],
                    res_fe_rA.params['l1_ln_x2_int'],
                    res_fe_l2.params['l2_ln_x2_int']],
    'EE': [res_fe.std_errors['l1_ln_x2_int'],
           res_fe_rA.std_errors['l1_ln_x2_int'],
           res_fe_l2.std_errors['l2_ln_x2_int']]
})
robust['IC_inf'] = robust['Coeficiente'] - 1.96*robust['EE']
robust['IC_sup'] = robust['Coeficiente'] + 1.96*robust['EE']

plt.figure(figsize=(6,4))
plt.errorbar(robust['Coeficiente'], robust['Modelo'],
             xerr=1.96*robust['EE'], fmt='o', capsize=3)
plt.axvline(0, linestyle='--')
plt.title("Robustez del coeficiente de intermodalidad")
plt.xlabel("Elasticidad estimada (IC95%)")
plt.tight_layout()
plt.savefig("grafico_robustez_x2.png", dpi=300)
plt.show()

#6. Distribución de residuos
plt.figure()
plt.hist(res_fe.resids, bins=30, color='gray', edgecolor='black')
plt.title("Distribución de residuos del modelo FE")
plt.xlabel("Residuos")
plt.ylabel("Frecuencia")
plt.tight_layout()
plt.savefig("grafico_residuos_FE.png", dpi=300)
plt.show()

# === Scatter UE-27: ln(Intensidad intermodal t−1) vs ln(PIB pc)
# === Siglas por país, colores distintos y escala fija “bonita”

import re
import numpy as np
import matplotlib.pyplot as plt
from itertools import cycle

# 1) (opcional) lista UE-27 para filtrar por si tu panel tuviera países extra
ue27 = {
    'Austria','Belgium','Bulgaria','Croatia','Cyprus','Czechia','Denmark','Estonia',
    'Finland','France','Germany','Greece','Hungary','Ireland','Italy','Latvia','Lithuania',
    'Luxembourg','Malta','Netherlands','Poland','Portugal','Romania','Slovakia','Slovenia',
    'Spain','Sweden'
}

# 2) columnas y helper para siglas
xcol, ycol = 'l1_ln_x2_int', 'ln_y'

def sigla_3(x):
    x = str(x).strip()
    # deja tal cual si ya viene como “ES”, “DE”, etc.; si no, genera A-Z (máx 3)
    return x.upper() if re.fullmatch(r'[A-Za-z]{2,3}', x) else re.sub(r'[^A-Za-z]', '', x)[:3].upper()

# 3) preparar datos
dfp = panel[['country','year',xcol,ycol]].dropna().copy()
# filtra a UE-27 si hace falta (no pasa nada si ya lo está)
dfp = dfp[dfp['country'].isin(ue27)] if len(ue27 - set(dfp['country'])) < len(ue27) else dfp
dfp['code'] = dfp['country'].map(sigla_3)

# 4) colores por país
countries = sorted(dfp['code'].unique())
palette = cycle(plt.get_cmap('tab20').colors)  # 20 colores; se reciclan si hay más
color_map = {c: next(palette) for c in countries}

# 5) figura
fig, ax = plt.subplots(figsize=(10,7))

# puntos por país
for c in countries:
    tmp = dfp.loc[dfp['code']==c].sort_values('year')
    ax.scatter(tmp[xcol], tmp[ycol], s=18, alpha=0.45, color=color_map[c])

# línea de tendencia global
X, Y = dfp[xcol].to_numpy(), dfp[ycol].to_numpy()
mask = np.isfinite(X) & np.isfinite(Y)
if mask.sum() >= 3:
    b1, b0 = np.polyfit(X[mask], Y[mask], 1)
    xx = np.linspace(np.nanmin(X), np.nanmax(X), 200)
    yy = b1*xx + b0
    ax.plot(xx, yy, '--', lw=2, alpha=0.85, color='gray', label='Tendencia global')

# etiquetas (siglas) en el centro típico (mediana) de cada país
centros = dfp.groupby(['code'])[[xcol, ycol]].median().reset_index()
for _, r in centros.iterrows():
    ax.text(r[xcol], r[ycol], r['code'],
            color=color_map[r['code']], fontsize=9,
            ha='center', va='center', alpha=0.95,
            fontweight='bold')

# 6) escala fija solicitada
ax.set_xlim(-8.1, -2.1)    # X: ln(Intensidad intermodal t−1)
ax.set_ylim(9.55, 11.50)   # Y: ln(PIB per cápita)

# 7) estética
ax.set_title("Relación entre intensidad intermodal rezagada y PIB per cápita (log) — UE-27")
ax.set_xlabel("ln(Intensidad intermodal t−1)")
ax.set_ylabel("ln(PIB per cápita)")
ax.grid(True, alpha=0.3)
# sin leyenda (las siglas ya identifican); si quieres, descomenta la siguiente línea:
# ax.legend(bbox_to_anchor=(1.02,1), loc='upper left', frameon=False)

plt.tight_layout()
plt.savefig("scatter_intermodalidad_PIB_UE27_siglas_colores.png", dpi=300, bbox_inches='tight')
plt.show()

```
